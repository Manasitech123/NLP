```python
# ================================================================
# NLP PRACTICAL 1
# 1. Perform tokenization (Whitespace, Punctuation-based, Treebank, Tweet, MWE) using NLTK library. 
# Use porter stemmer and snowball stemmer for stemming. Use any technique for lemmatization.  
# ================================================================

# ================================================================
# INSTALLATION COMMANDS
# ================================================================
#
# pip install nltk
#
# ------------------------------------------------
# Download Required NLTK Resources (Run Once)
# ------------------------------------------------
#
# python
#
# import nltk
# nltk.download('punkt')
# nltk.download('wordnet')
# nltk.download('omw-1.4')
# nltk.download('averaged_perceptron_tagger')
# nltk.download('averaged_perceptron_tagger_eng')
#
# exit()
#
# ================================================================

# ================================================================
# IMPORT LIBRARIES
# ================================================================

import nltk

from nltk.tokenize import (
    WhitespaceTokenizer,
    PunktSentenceTokenizer,
    TreebankWordTokenizer,
    WordPunctTokenizer,
    TweetTokenizer,
    MWETokenizer,
    word_tokenize
)

from nltk.stem import (
    PorterStemmer,
    SnowballStemmer,
    WordNetLemmatizer
)

from nltk.corpus import wordnet

# ================================================================
# FUNCTION : CONVERT POS TAGS
# ================================================================

def get_wordnet_pos(word):

    # Get POS tag
    tag = nltk.pos_tag([word])[0][1][0].upper()

    tag_dict = {
        "J": wordnet.ADJ,
        "N": wordnet.NOUN,
        "V": wordnet.VERB,
        "R": wordnet.ADV
    }

    return tag_dict.get(tag, wordnet.NOUN)

# ================================================================
# SYSTEM HEADER
# ================================================================

print("\n" + "="*70)
print(" TOKENIZATION, STEMMING AND LEMMATIZATION SYSTEM")
print("="*70)

# ================================================================
# USER INPUT
# ================================================================

text = input("\nEnter Sentence : ")

# Convert text to lowercase
text_lower = text.lower()

# ================================================================
# DISPLAY ORIGINAL TEXT
# ================================================================

print("\n" + "="*70)
print(" ORIGINAL TEXT")
print("="*70)

print(text)

# ================================================================
# TOKENIZATION
# ================================================================

print("\n" + "="*70)
print(" TOKENIZATION OUTPUT")
print("="*70)

# ------------------------------------------------
# 1. Whitespace Tokenization
# ------------------------------------------------

wt = WhitespaceTokenizer()

print("\n1. Whitespace Tokenization")
print("-"*40)

print(wt.tokenize(text_lower))

# ------------------------------------------------
# 2. Punctuation-based Tokenization
# ------------------------------------------------

# ------------------------------------------------
# 2. Punctuation-based Tokenization
# ------------------------------------------------

pt = WordPunctTokenizer()

print("\n2. Punctuation-based Tokenization")
print("-"*40)

print(pt.tokenize(text_lower))

# ------------------------------------------------
# 3. Treebank Tokenization
# ------------------------------------------------

treebank = TreebankWordTokenizer()

print("\n3. Treebank Tokenization")
print("-"*40)

print(treebank.tokenize(text_lower))

# ------------------------------------------------
# 4. Tweet Tokenization
# ------------------------------------------------

tweet = TweetTokenizer()

print("\n4. Tweet Tokenization")
print("-"*40)

print(tweet.tokenize(text_lower))

# ------------------------------------------------
# 5. MWE Tokenization
# ------------------------------------------------

mwe = MWETokenizer([('machine', 'learning')], separator='_')

print("\n5. MWE Tokenization")
print("-"*40)

tokens = word_tokenize(text_lower)

print(mwe.tokenize(tokens))

# ================================================================
# STEMMING
# ================================================================

words = word_tokenize(text_lower)

porter = PorterStemmer()

snowball = SnowballStemmer("english")

print("\n" + "="*70)
print(" STEMMING OUTPUT")
print("="*70)

# ------------------------------------------------
# Porter Stemmer
# ------------------------------------------------

print("\n1. Porter Stemmer")
print("-"*40)

for word in words:

    print(f"{word:<20} --> {porter.stem(word)}")

# ------------------------------------------------
# Snowball Stemmer
# ------------------------------------------------

print("\n2. Snowball Stemmer")
print("-"*40)

for word in words:

    print(f"{word:<20} --> {snowball.stem(word)}")

# ================================================================
# LEMMATIZATION
# ================================================================

lemmatizer = WordNetLemmatizer()

print("\n" + "="*70)
print(" LEMMATIZATION OUTPUT")
print("="*70)

for word in words:

    # Get grammatical category
    pos = get_wordnet_pos(word)

    # Perform proper lemmatization
    lemma = lemmatizer.lemmatize(word, pos)

    print(f"{word:<20} --> {lemma}")

# ================================================================
# PROGRAM END
# ================================================================

print("\n" + "="*70)
print(" PROGRAM EXECUTED SUCCESSFULLY")
print("="*70)


#I am learning machine learning! NLP is amazing :) #AI @OpenAI
```

# Theory

Natural Language Processing (NLP) is a branch of Artificial Intelligence that enables computers to understand, interpret, and process human language. Before text data can be used for machine learning or language analysis, it must undergo preprocessing. Tokenization, stemming, and lemmatization are important preprocessing techniques used in NLP.

---

# 1. Tokenization

Tokenization is the process of breaking text into smaller units called tokens. These tokens may be words, sentences, or phrases.

## Types of Tokenization Used

### a) Whitespace Tokenization

Splits text based on spaces.

#### Example

`Machine Learning is powerful`

#### Output

`['Machine', 'Learning', 'is', 'powerful']`

---

### b) Punctuation-based Tokenization

Splits text into sentences or tokens based on punctuation symbols.

---

### c) Treebank Tokenization

A rule-based tokenizer designed for English text. It handles contractions and punctuation efficiently.

---

### d) Tweet Tokenization

Special tokenizer optimized for social media text such as hashtags, mentions, emojis, and abbreviations.

---

### e) Multi-Word Expression (MWE) Tokenization

Combines multiple words into a single token.

#### Example

`machine learning → machine_learning`

---

# 2. Stemming

Stemming is the process of reducing words to their root or stem form by removing suffixes.

#### Example

`playing → play`

`studies → studi`

---

## Types of Stemmers

### Porter Stemmer

A commonly used stemming algorithm that removes suffixes using predefined linguistic rules.

---

### Snowball Stemmer

An improved version of Porter Stemmer with better accuracy and support for multiple languages.

---

# 3. Lemmatization

Lemmatization converts words into their meaningful dictionary base form called a lemma.

Unlike stemming, lemmatization considers:
- Grammar
- Context
- Part of Speech

#### Example

`children → child`

`playing → play`

`are → be`

In this practical, WordNet Lemmatizer along with Part-of-Speech tagging is used for accurate lemmatization.
