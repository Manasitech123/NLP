````python
# ================================================================
#  3. Perform text cleaning, perform lemmatization (any method), remove stop words (any method), label encoding. 
# Create representations using TF-IDF. Save outputs. 
# ================================================================

# ================================================================
# INSTALLATION COMMANDS
# ================================================================
#
# pip install nltk
# pip install pandas
# pip install scikit-learn
#
# ------------------------------------------------
# Download NLTK Resources (Run Once)
# ------------------------------------------------
#
# python
#
# import nltk
# nltk.download('punkt')
# nltk.download('wordnet')
# nltk.download('omw-1.4')
# nltk.download('stopwords')
#
# exit()
#
# ================================================================

# ================================================================
# IMPORT LIBRARIES
# ================================================================

import re
import pandas as pd
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer

# ================================================================
# SYSTEM HEADER
# ================================================================

print("\n" + "="*70)
print("            NLP TEXT PROCESSING AND TF-IDF SYSTEM")
print("="*70)

# ================================================================
# USER INPUT
# ================================================================

n = int(input("\nEnter Number of Sentences : "))

text_data = []
labels = []

print("\n" + "="*70)
print("ENTER SENTENCES AND LABELS")
print("="*70)

for i in range(n):

    print(f"\nSentence {i+1}")

    sentence = input("Enter Text   : ")

    label = input("Enter Label  : ").lower()

    text_data.append(sentence)
    labels.append(label)

# ================================================================
# DISPLAY ORIGINAL DATA
# ================================================================

print("\n" + "="*70)
print("ORIGINAL INPUT DATA")
print("="*70)

for i in range(n):

    print(f"\nRecord {i+1}")
    print("-"*40)
    print(f"Text   : {text_data[i]}")
    print(f"Label  : {labels[i]}")

# ================================================================
# TEXT CLEANING
# ================================================================

stop_words = set(stopwords.words('english'))

cleaned_data = []

for text in text_data:

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation and numbers
    text = re.sub(r'[^a-z\s]', '', text)

    # Tokenization
    words = word_tokenize(text)

    # Remove stopwords
    words = [word for word in words if word not in stop_words]

    cleaned_data.append(words)

# ================================================================
# DISPLAY CLEANED DATA
# ================================================================

print("\n" + "="*70)
print("TEXT CLEANING OUTPUT")
print("="*70)

for i, sentence in enumerate(cleaned_data):

    print(f"\nSentence {i+1}")
    print("-"*40)
    print("Tokens :", sentence)

# ================================================================
# LEMMATIZATION
# ================================================================

lemmatizer = WordNetLemmatizer()

lemmatized_data = []

for sentence in cleaned_data:

    temp = []

    for word in sentence:

        lemma = lemmatizer.lemmatize(word)

        temp.append(lemma)

    final_sentence = " ".join(temp)

    lemmatized_data.append(final_sentence)

# ================================================================
# DISPLAY LEMMATIZED DATA
# ================================================================

print("\n" + "="*70)
print("LEMMATIZATION OUTPUT")
print("="*70)

for i, sentence in enumerate(lemmatized_data):

    print(f"\nSentence {i+1}")
    print("-"*40)
    print(sentence)

# ================================================================
# LABEL ENCODING
# ================================================================

encoder = LabelEncoder()

encoded_labels = encoder.fit_transform(labels)

# ================================================================
# DISPLAY LABEL ENCODING
# ================================================================

print("\n" + "="*70)
print("LABEL ENCODING")
print("="*70)

for original, encoded in zip(labels, encoded_labels):

    print(f"{original:<15} --> {encoded}")

# ================================================================
# TF-IDF REPRESENTATION
# ================================================================

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(lemmatized_data)

# ================================================================
# DISPLAY TF-IDF VOCABULARY
# ================================================================

print("\n" + "="*70)
print("TF-IDF VOCABULARY")
print("="*70)

for word in tfidf.get_feature_names_out():

    print(word)

# ================================================================
# DISPLAY TF-IDF MATRIX
# ================================================================

print("\n" + "="*70)
print("TF-IDF MATRIX")
print("="*70)

matrix = tfidf_matrix.toarray()

for i, row in enumerate(matrix):

    print(f"\nSentence {i+1}")
    print("-"*40)

    for word, value in zip(tfidf.get_feature_names_out(), row):

        print(f"{word:<15} : {value:.4f}")

# ================================================================
# CREATE DATAFRAME
# ================================================================

df = pd.DataFrame(
    matrix,
    columns=tfidf.get_feature_names_out()
)

df["label"] = encoded_labels

# ================================================================
# SAVE OUTPUT
# ================================================================

df.to_csv("tfidf_output.csv", index=False)

# ================================================================
# FINAL OUTPUT
# ================================================================

print("\n" + "="*70)
print("FINAL DATAFRAME")
print("="*70)

print(df)

print("\n" + "="*70)
print("OUTPUT FILE GENERATED SUCCESSFULLY")
print("="*70)

print("\nSaved File Name : tfidf_output.csv")

print("\nProgram Executed Successfully.")
print("="*70)

#Text                                  Label
#------------------------------------------------
#I love machine learning               Positive
#This movie is terrible                Negative
#NLP is very interesting               Positive
#I hate spam emails                    Negative
````

# Theory

Natural Language Processing (NLP) is a branch of Artificial Intelligence that enables computers to understand, interpret, and process human language. Raw textual data often contains unwanted symbols, punctuation, stopwords, and inconsistent word forms. Therefore, preprocessing techniques such as text cleaning, stopword removal, lemmatization, label encoding, and TF-IDF representation are essential before applying machine learning algorithms.

These preprocessing techniques help convert unstructured text into structured numerical representations suitable for machine learning and deep learning applications.

---

# 1. Text Cleaning

Text cleaning is the process of removing unnecessary and unwanted elements from textual data to improve text quality and consistency.

The following operations are generally performed during text cleaning:

- Converting text to lowercase
- Removing punctuation marks
- Removing numbers
- Removing special characters
- Removing extra spaces

## Example

Original Text:

`I Love NLP!!!`

Cleaned Text:

`i love nlp`

Text cleaning reduces noise in the dataset and improves the performance of NLP models.

---

# 2. Tokenization

Tokenization is the process of breaking text into smaller units called tokens. These tokens may be words, sentences, or phrases.

## Example

Sentence:

`Natural Language Processing is interesting`

Output Tokens:

`['Natural', 'Language', 'Processing', 'is', 'interesting']`

In this practical, tokenization is performed using the `word_tokenize()` function from the NLTK library.

---

# 3. Stopword Removal

Stopwords are commonly used words that do not contribute significant meaning in text analysis.

Examples:
- is
- the
- am
- are
- and

Removing stopwords:
- reduces dataset size
- improves processing speed
- enhances model efficiency

## Example

Original Sentence:

`NLP is very interesting`

After Stopword Removal:

`NLP interesting`

In this practical, stopwords are removed using the NLTK stopword corpus.

---

# 4. Lemmatization

Lemmatization is the process of converting words into their meaningful dictionary base form called a lemma.

Unlike stemming, lemmatization considers:
- grammar
- context
- part of speech (POS)

## Example

| Original Word | Lemmatized Word |
|---|---|
| children | child |
| playing | play |
| studies | study |
| are | be |

Lemmatization improves text consistency while preserving proper linguistic meaning.

In this practical:
- `WordNetLemmatizer` is used for lemmatization.
- POS tagging is used for accurate grammatical interpretation.

---

# 5. Part-of-Speech (POS) Tagging

POS tagging identifies the grammatical category of a word such as:
- noun
- verb
- adjective
- adverb

POS tagging improves lemmatization accuracy.

## Example

| Word | POS Tag |
|---|---|
| playing | Verb |
| beautiful | Adjective |
| quickly | Adverb |

In this practical, POS tagging is performed using the `nltk.pos_tag()` function.

---

# 6. Label Encoding

Machine learning algorithms cannot process categorical labels directly. Therefore, labels such as:

- positive
- negative

must be converted into numerical values.

This process is called Label Encoding.

## Example

| Original Label | Encoded Value |
|---|---|
| negative | 0 |
| positive | 1 |

In this practical, `LabelEncoder` from Scikit-learn is used for label encoding.

---

# 7. TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF is a statistical technique used to measure the importance of a word in a document relative to a collection of documents.

It combines:
- Term Frequency (TF)
- Inverse Document Frequency (IDF)

---

## Term Frequency (TF)

TF measures how frequently a word appears in a document.

### Formula

TF = (Number of times term appears in document) / (Total number of terms in document)

---

## Inverse Document Frequency (IDF)

IDF measures how unique or rare a word is across all documents.

### Formula

IDF = log(Total Number of Documents / Number of Documents containing the term)

---

## TF-IDF Formula

TF-IDF = TF × IDF

Words with high TF-IDF scores are considered more important in the document.

TF-IDF is widely used in:
- Search engines
- Information retrieval
- Text classification
- Sentiment analysis
- Document ranking
