```python
# ==========================================
# 5. Morphology is the study of the way words are built up from smaller meaning bearing units. Study 
#    and understand the concepts of morphology by the use of add delete table 
# ==========================================

# Morphology means:
# Study of word formation using prefixes and suffixes.

# ==========================================
# IMPORT LIBRARY
# ==========================================

import pandas as pd

# ==========================================
# VALID PREFIXES AND SUFFIXES
# ==========================================

# Only meaningful combinations are included

valid_prefixes = {
    "happy": ["un"],
    "agree": ["dis"],
    "play": ["re"],
    "connect": ["dis"]
}

valid_suffixes = {
    "play": ["er", "ing", "ful"],
    "happy": ["ness"],
    "care": ["ful", "less"],
    "develop": ["ment"],
    "use": ["ful"],
    "agree": ["ment"]
}

# ==========================================
# FUNCTION TO ADD AFFIXES
# ==========================================

def add_affixes(word):

    results = []

    # Add Valid Prefixes
    if word in valid_prefixes:

        for prefix in valid_prefixes[word]:

            results.append(
                (word, "+" + prefix, prefix + word)
            )

    # Add Valid Suffixes
    if word in valid_suffixes:

        for suffix in valid_suffixes[word]:

            # Grammar Rule
            if word.endswith("y") and suffix == "ness":

                new_word = word[:-1] + "iness"

            else:

                new_word = word + suffix

            results.append(
                (word, "+" + suffix, new_word)
            )

    return results

# ==========================================
# FUNCTION TO DELETE AFFIXES
# ==========================================

def delete_affixes(word):

    results = []

    # Prefix Deletion
    prefixes = ["un", "re", "dis"]

    for prefix in prefixes:

        if word.startswith(prefix):

            root = word[len(prefix):]

            results.append(
                (word, "-" + prefix, root)
            )

    # Suffix Deletion
    suffixes = [
        "er",
        "ing",
        "ness",
        "less",
        "ful",
        "ment"
    ]

    for suffix in suffixes:

        if word.endswith(suffix):

            root = word[:-len(suffix)]

            # Grammar correction
            if suffix == "ness" and root.endswith("i"):

                root = root[:-1] + "y"

            results.append(
                (word, "-" + suffix, root)
            )

    return results

# ==========================================
# MAIN PROGRAM
# ==========================================

print("\n" + "="*70)
print("        MORPHOLOGY USING ADD-DELETE TABLE")
print("="*70)

# User Input
word = input("\nEnter a Base Word : ").lower()

# Validation
if not word.isalpha():

    print("\nPlease Enter a Valid Alphabetic Word.")

else:

    # ==========================================
    # ADD TABLE
    # ==========================================

    print("\n" + "="*70)
    print("ADD TABLE")
    print("="*70)

    add_results = add_affixes(word)

    if len(add_results) == 0:

        print("\nNo Valid Morphological Additions Found.")

    else:

        add_df = pd.DataFrame(
            add_results,
            columns=["Base Word", "Operation", "New Word"]
        )

        print(add_df.to_string(index=False))

    # ==========================================
    # DELETE TABLE
    # ==========================================

    print("\n" + "="*70)
    print("DELETE TABLE")
    print("="*70)

    delete_results = delete_affixes(word)

    if len(delete_results) == 0:

        print("\nNo Prefixes or Suffixes Found for Deletion.")

    else:

        delete_df = pd.DataFrame(
            delete_results,
            columns=["Word", "Operation", "Root Word"]
        )

        print(delete_df.to_string(index=False))

    # ==========================================
    # SAVE OUTPUT
    # ==========================================

    if len(add_results) != 0:
        add_df.to_csv("morphology_add_table.csv", index=False)

    if len(delete_results) != 0:
        delete_df.to_csv("morphology_delete_table.csv", index=False)

    # ==========================================
    # SUCCESS MESSAGE
    # ==========================================

    print("\n" + "="*70)
    print("OUTPUT FILES GENERATED SUCCESSFULLY")
    print("="*70)

    if len(add_results) != 0:
        print("1. morphology_add_table.csv")

    if len(delete_results) != 0:
        print("2. morphology_delete_table.csv")

    print("\nProgram Executed Successfully.")
    print("="*70)

#happy, play, write, teach
```

# Theory

Morphology is a branch of Natural Language Processing (NLP) and Linguistics that studies the internal structure and formation of words. It analyzes how words are created using smaller meaningful units called morphemes.

A morpheme is the smallest meaningful unit in a language. Words can be formed by combining:
- Root words
- Prefixes
- Suffixes

Morphological analysis helps computers understand word structure, grammatical variations, and word transformations.

---

# Morphology in NLP

In Natural Language Processing, morphology is used to:
- analyze word structures
- identify root words
- remove prefixes and suffixes
- generate new words
- improve text normalization

Morphology plays an important role in:
- search engines
- machine translation
- spell checking
- information retrieval
- chatbots
- sentiment analysis

---

# Morphemes

A morpheme is the smallest unit of meaning in a word.

## Example

| Word | Morphemes |
|---|---|
| unhappy | un + happy |
| playing | play + ing |
| disagreement | dis + agree + ment |

---

# Types of Morphemes

## 1. Root Word

The main base word that carries the core meaning.

### Example

| Word | Root Word |
|---|---|
| playing | play |
| happiness | happy |
| disagreement | agree |

---

## 2. Prefix

A prefix is added at the beginning of a word to modify its meaning.

### Examples

| Prefix | Word | Meaning |
|---|---|---|
| un | unhappy | not happy |
| re | replay | play again |
| dis | disagree | opposite meaning |

---

## 3. Suffix

A suffix is added at the end of a word to change its grammatical form or meaning.

### Examples

| Suffix | Word | Meaning |
|---|---|---|
| ing | playing | continuous action |
| er | player | person performing action |
| ness | happiness | state or quality |

---

# Add-Delete Table

An Add-Delete Table is a morphology technique used to study how words are formed by adding or removing prefixes and suffixes.

It contains two operations:

## 1. Add Operation

Adds prefixes or suffixes to a root word.

### Example

| Base Word | Operation | Result |
|---|---|---|
| play | +ing | playing |
| happy | +ness | happiness |
| play | +re | replay |

---

## 2. Delete Operation

Removes prefixes or suffixes from a word to obtain the root word.

### Example

| Word | Operation | Root Word |
|---|---|---|
| replay | -re | play |
| happiness | -ness | happy |
| playing | -ing | play |

---

# Inflectional and Derivational Morphology

Morphology is mainly divided into two categories:

---

## 1. Inflectional Morphology

Inflectional morphology changes the grammatical form of a word without changing its core meaning.

### Examples

| Original Word | Inflected Word |
|---|---|
| play | played |
| walk | walking |
| cat | cats |

Characteristics:
- Does not create a new word
- Changes tense, number, or form

---

## 2. Derivational Morphology

Derivational morphology creates new words with new meanings.

### Examples

| Root Word | Derived Word |
|---|---|
| happy | unhappy |
| play | player |
| agree | disagreement |

Characteristics:
- Creates new vocabulary
- Changes meaning or word class
