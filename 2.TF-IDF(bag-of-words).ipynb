```python
# ================================================================
# NLP PRACTICAL 2
# 2. Perform bag-of-words approach (count occurrence, normalized count occurrence), TF-IDF on data.Create embeddings using Word2Vec. 
# ================================================================

# ================================================================
# INSTALLATION COMMANDS
# ================================================================
#
# pip install pandas
# pip install numpy
# pip install scikit-learn
# pip install gensim
#
# ================================================================

# ================================================================
# IMPORT LIBRARIES
# ================================================================

import pandas as pd

from sklearn.feature_extraction.text import (
    CountVectorizer,
    TfidfVectorizer
)

from gensim.models import Word2Vec

# ================================================================
# SYSTEM HEADER
# ================================================================

print("\n" + "="*75)
print("         BAG OF WORDS, TF-IDF AND WORD2VEC SYSTEM")
print("="*75)

# ================================================================
# USER INPUT
# ================================================================

n = int(input("\nEnter Number of Sentences : "))

corpus = []

print("\n" + "="*75)
print("ENTER SENTENCES")
print("="*75)

for i in range(n):

    sentence = input(f"\nEnter Sentence {i+1} : ")

    corpus.append(sentence)

# ================================================================
# DISPLAY INPUT CORPUS
# ================================================================

print("\n" + "="*75)
print("INPUT CORPUS")
print("="*75)

for i, sentence in enumerate(corpus):

    print(f"Sentence {i+1} : {sentence}")

# ================================================================
# BAG OF WORDS
# ================================================================

cv = CountVectorizer()

bow_matrix = cv.fit_transform(corpus)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=cv.get_feature_names_out()
)

print("\n" + "="*75)
print("BAG OF WORDS MATRIX")
print("="*75)

print(bow_df)

# ================================================================
# COUNT OCCURRENCE
# ================================================================

combined_doc = " ".join(corpus)

vectorizer = CountVectorizer()

counts = vectorizer.fit_transform([combined_doc])

print("\n" + "="*75)
print("COUNT OCCURRENCE")
print("="*75)

for word, count in zip(
    vectorizer.get_feature_names_out(),
    counts.toarray()[0]
):

    print(f"{word:<20} : {count}")

# ================================================================
# NORMALIZED COUNT OCCURRENCE
# ================================================================

# Formula:
# Normalized Count = Word Count / Total Words

normalized_bow = bow_df.div(
    bow_df.sum(axis=1),
    axis=0
)

print("\n" + "="*75)
print("NORMALIZED COUNT OCCURRENCE")
print("="*75)

print(normalized_bow)

# ================================================================
# TF-IDF REPRESENTATION
# ================================================================

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(corpus)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("\n" + "="*75)
print("TF-IDF MATRIX")
print("="*75)

print(tfidf_df)

# ================================================================
# WORD2VEC EMBEDDINGS
# ================================================================

# Convert sentences into tokenized format
tokenized_sentences = []

for sentence in corpus:

    tokens = sentence.lower().split()

    tokenized_sentences.append(tokens)

# Create Word2Vec Model
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=5,
    window=2,
    min_count=1,
    workers=1
)

print("\n" + "="*75)
print("WORD2VEC EMBEDDINGS")
print("="*75)

for word in model.wv.index_to_key:

    print(f"\nWord : {word}")
    print("-"*40)

    print(model.wv[word])

# ================================================================
# SAVE OUTPUTS
# ================================================================

bow_df.to_csv("bow_output.csv", index=False)

tfidf_df.to_csv("tfidf_output.csv", index=False)

normalized_bow.to_csv(
    "normalized_count_output.csv",
    index=False
)

print("\n" + "="*75)
print("OUTPUT FILES GENERATED SUCCESSFULLY")
print("="*75)

print("\nSaved Files :")
print("1. bow_output.csv")
print("2. tfidf_output.csv")
print("3. normalized_count_output.csv")

# ================================================================
# PROGRAM END
# ================================================================

print("\n" + "="*75)
print("PROGRAM EXECUTED SUCCESSFULLY")
print("="*75)


#Enter Sentence 1 : Machine learning is fun
#Enter Sentence 2 : Deep learning is powerful
#Enter Sentence 3 : Natural language processing uses machine learning
```

# Theory

Natural Language Processing (NLP) systems cannot directly process textual information in its raw form. Therefore, text data must be converted into numerical representations before applying machine learning algorithms. Various text representation techniques such as Bag of Words (BoW), Count Occurrence, Normalized Count Occurrence, TF-IDF, and Word2Vec are commonly used for transforming text into meaningful vector formats. These techniques help machines analyze, understand, and process human language efficiently.

## Bag of Words (BoW)

Bag of Words is one of the simplest and most widely used text representation techniques in NLP. In this method, a document is represented as a collection of words without considering grammar, sentence structure, or word order. Each unique word in the dataset becomes a feature, and the value assigned represents the number of times that word appears in the document.

For example, if the sentences are:

`Data science is fun`

`Science is interesting`

the vocabulary becomes:

`[data, science, is, fun, interesting]`

Each sentence is then represented as a vector of word frequencies. Bag of Words is simple and effective for many text classification problems, but it does not capture semantic meaning or relationships between words.

---

## Count Occurrence

Count occurrence refers to counting how many times each word appears in a document or corpus. It is the foundation of the Bag of Words model. Words with higher frequency may indicate greater importance in the document.

Example:

- science : 3  
- data : 2  
- fun : 2  

This method is easy to implement and useful for identifying frequently occurring terms, but it may give excessive importance to commonly used words.

---

## Normalized Count Occurrence

Normalized count occurrence improves upon simple frequency counting by dividing the count of each word by the total number of words in the document. This provides relative frequency instead of absolute frequency.

### Formula

Normalized Count = Word Frequency / Total Number of Words

Normalization helps reduce bias caused by document length and provides a more balanced comparison between documents of different sizes.

---

## TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF is an advanced statistical technique used to evaluate the importance of a word in a document relative to an entire collection of documents. It combines two measures: Term Frequency (TF) and Inverse Document Frequency (IDF).

### Term Frequency (TF)

TF measures how often a word appears in a document.

### Formula

TF = (Number of times term appears in document) / (Total number of terms in document)

---

### Inverse Document Frequency (IDF)

IDF measures how unique or rare a word is across all documents.

### Formula

IDF = log(Total Number of Documents / Number of Documents containing the term)

---

### TF-IDF Formula

TF-IDF = TF × IDF

TF-IDF gives higher weight to important words that occur frequently in a specific document but less frequently across the entire corpus. Common words such as “is”, “the”, and “and” usually receive lower weights.

TF-IDF is widely used in:
- Search engines
- Document ranking
- Text mining
- Information retrieval
- Sentiment analysis

## Word2Vec

Word2Vec is a neural network-based word embedding technique developed by Google. Unlike Bag of Words and TF-IDF, Word2Vec captures semantic meaning and contextual relationships between words.

In Word2Vec:
- Similar words have similar vector representations.
- Words are converted into dense numerical vectors.
- Semantic relationships between words are preserved.

Example:

`king - man + woman ≈ queen`

Word2Vec learns word relationships by analyzing surrounding words in a sentence. It uses shallow neural networks to generate embeddings and is highly effective for tasks involving semantic understanding.

There are two major Word2Vec architectures:
1. CBOW (Continuous Bag of Words)
2. Skip-Gram Model

### Advantages of Word2Vec
- Captures semantic meaning
- Handles contextual similarity
- Generates dense vector representations
- Improves NLP model performance

### Applications of Word2Vec
- Chatbots
- Recommendation systems
- Machine translation
- Sentiment analysis
- Search engines
- Text summarization

Overall, Bag of Words, TF-IDF, and Word2Vec are essential text representation techniques used in Natural Language Processing. They help convert unstructured text into structured numerical data suitable for machine learning and deep learning applications.
